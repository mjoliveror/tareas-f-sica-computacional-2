```python
#2)
a_t = np.linspace(0,50,500)

flujo = np.ones(500)

#ruido gaussiano
ruido = rgn.normal(loc=0.0, scale=0.002, size= 500)
flujo += ruido

mascara_transito = (a_t >= 4.5) & (a_t <= 5.0)
flujo[mascara_transito] -= 0.02

plt.hist(flujo)
```

```python
#a)
flujo_prom = np.mean(flujo)
print(flujo_prom)
```

```python
#b)
flujo_des_est = np.std(flujo)
print(flujo_des_est)
```

```python
#c)
flujo_min = np.min(flujo)
print(flujo_min)
# este es el indice
np.where(flujo > flujo_min)
```

```python
#e)
elementos_que_cumplen = flujo_prom - 3 * flujo_des_est
criterio = flujo < elementos_que_cumplen

#lo que sí cumple
puntos_que_cumplen = flujo[criterio]
print(puntos_que_cumplen)
```

```python
#f)
posibles_puntos = np.sum(criterio)
print(posibles_puntos)
```

```python
#g)
transito = np.mean(puntos_que_cumplen)
delta_f = 1- transito
print(delta_f)
```

```python
#h)
razon_radios = np.sqrt(delta_f)
print(razon_radios)
```
