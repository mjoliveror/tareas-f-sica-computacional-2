```python
#3) Función para calcular la energía cinética
def energía_cinética(m,v_kmh):
  v_ms = v_kmh * (1000/3600)

  ec = 0.5 * m * v_ms ** 2

  return ec

masa_kg = 4
velocidad_kmh = 2

resultado = energía_cinética(masa_kg, velocidad_kmh)

print(f'la energía cinetica es: {resultado}')
```
